### Q1. CNN, RNN/LSTM and GAN Concept Matching

#### Concept Matching Table

| Model | Suitable Data | Cyber-Security Example |
| :--- | :--- | :--- |
| **CNN** | Images / grid-like data | Malware image classification or packet-payload matrix visualization |
| **RNN / LSTM** | Sequences / logs / time series | Sequential intrusion detection or network traffic authentication-log analysis |
| **GAN** | Synthetic-data generation | Generating controlled synthetic cyber-attack samples for data augmentation |
| **Dense Network** | Small numeric feature table | Normal / Attack binary classification based on activity metrics (as done in this lab) |

#### Why is a Dense Network used in the practical implementation of this lab?

A dense network is used because the input dataset (`simple_cyber_activity.csv`) consists of a small set of structured, independent numerical features (`failed_logins`, `login_hour`, `new_device`, and `data_transfer_mb`). There is no inherent spatial grid-like structure (which would require a CNN) or sequence/temporal dependencies (which would require an RNN/LSTM) in these independent tabular attributes.

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv('simple_cyber_activity.csv')

# Select features
features = [
    'failed_logins',
    'login_hour',
    'new_device',
    'data_transfer_mb'
]

# Define input (X) and output (y)
X = df[features]
y = df['label']

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Apply feature scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Display dataset shapes
print('Training shape:', X_train_scaled.shape)
print('Testing shape:', X_test_scaled.shape)
```

```python
import tensorflow as tf
import numpy as np

# Set random seeds
np.random.seed(42)
tf.random.set_seed(42)

# Create Sequential Neural Network
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(4, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Display model architecture
model.summary()

# Train the model
history = model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=20,
    batch_size=8,
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

# Plot training and validation accuracy
plt.plot(
    history.history['accuracy'],
    label='Training accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation accuracy'
)

plt.title('Neural Network Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()


# Plot training and validation loss
plt.plot(
    history.history['loss'],
    label='Training loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation loss'
)

plt.title('Neural Network Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()


# Evaluate model on test dataset
test_loss, test_accuracy = model.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

# Display results
print(f'Test accuracy: {test_accuracy * 100:.2f}%')
print(f'Test loss: {test_loss:.4f}')
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Create new cyber activity records
new_records = pd.DataFrame([
    {
        'failed_logins': 0,
        'login_hour': 11,
        'new_device': 0,
        'data_transfer_mb': 35
    },
    {
        'failed_logins': 7,
        'login_hour': 1,
        'new_device': 1,
        'data_transfer_mb': 250
    },
    {
        'failed_logins': 2,
        'login_hour': 23,
        'new_device': 1,
        'data_transfer_mb': 850
    }
])

# Scale new records
new_scaled = scaler.transform(new_records)

# Neural Network predictions
probabilities = model.predict(
    new_scaled,
    verbose=0
).flatten()

# Add neural network results
new_records['nn_probability'] = probabilities.round(3)

new_records['nn_prediction'] = np.where(
    probabilities >= 0.50,
    'Attack',
    'Normal'
)

# Create Decision Tree model
tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

# Train Decision Tree
tree.fit(X_train, y_train)

# Predict test data
tree_pred = tree.predict(X_test)

# Calculate Decision Tree accuracy
tree_accuracy = accuracy_score(
    y_test,
    tree_pred
)

# Predict new records using Decision Tree
new_records['tree_prediction'] = np.where(
    tree.predict(new_records[features]) == 1,
    'Attack',
    'Normal'
)

# Display model accuracy
print(
    f'Neural-network accuracy: {test_accuracy * 100:.2f}%'
)

print(
    f'Decision-tree accuracy: {tree_accuracy * 100:.2f}%'
)

# Display predictions
display(new_records)
```
